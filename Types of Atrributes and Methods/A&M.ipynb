```python
class Animal:
    a = 12 # Class Attribute

    def __init__(self, name):
        self.name = name   # Object / Instance attribute

    def hello(self):       # Instance / Object Method. Captures the location of object
        print(f"My name is {self.name}")

    @classmethod           # Decorator
    def importance(cls):   # Class method. Captures the location of Class
        print("Yo!")

    @staticmethod
    def speak():           # This is a static method and it will not target any location.
        print("What's up!")    

obj = Animal("Lion") 

print(obj.name)

obj.importance()
```
