```python
# If we want to take input from user we cannot do that using 
# arguments in a class. For that we use a constructor.

# A constructor is a method that runs automatically when we 
# call a class and this constructor function will target the
# objects location.

class Bags:
    def __init__(self, materials, zips, pockets):
        self.materials = materials
        self.zips = zips
        self.pockets = pockets
# self.x ka use tab aayega when we have multiple objects.
# Har object ke andar different arguments pass hogi
# Toh it is these guys jobs to take the material given to them and store it inside this object.         
# Self ke andar object ki location jaati hai it doesn't need any input.

reebok = Bags("Leather", 3, 4)
campus = Bags("Nylon", 2, 2)

print(reebok.materials)
print(campus.materials)
```
