```python
# Inheritance allows a class (child class) to inherit properties and behaviors (attributes and methods) from another class (parent class).
# Benefits of using inheritance is:
# > Code Reusability
# > Organised structure
# > Easy to maintain and extend

class Parent:
    x = 10
    def __init__(self, name):
        self.name = name
        print("I can speak.")

    def importance(self):
        print(f"Hello your name is {self.name}.")    

class Child(Parent):   # This is the syntax of a class just that the parameters that you take are of the parent class
    pass

obj = Parent("gupta")
obj1 = Child("Dhirender")

obj1.importance()
print(obj1.x)

# Your child class has all the powers to access the attributes and methods of parent class.
```

```python
# Constructor in Inheritance

class BagFactory:
    def __init__(self, material, zips, pockets):
        self.material = material
        self.zips = zips
        self.pockets = pockets

    def details(self):
        print("Your Bag details are: ")    
        print("Material: ", self.material)
        print("Zips: ", self.zips)
        print("Pockets: ", self.pockets)

class Reebok(BagFactory):     # Reebok is a child class of BagFactory but can still add things of its own.
    def __init__(self, material, zips, pockets, colours):
        super().__init__(material, zips, pockets) # Think of super as let the parent class do thier constructor first and then the child class will do the extra work afterwards.
        self.colours = colours

    """
     What we are doing here is:
     First we created another details function so that we can the new detail which is "colour" in the output by using print.
     Next we are calling the details function of the parent class (BagFactory) so basically this is the simple flowchart of what this code is doing.

     Reebok.details()
           ↓
     Print colour
           ↓
     Call BagFactory.details()
           ↓
     Print material, zips, pockets
    """

    def details(self):
        print("Colour: ", self.colours) 
        return super().details()    

bag1 = BagFactory("Leather", 3, 2) 

bag2 = Reebok("Nylon", 2, 1, "Green")

bag1.details()
bag2.details()

# This type of inheritance is called Single type inheritance.
# Definition: One child class inherits from one parent class.
```

```python
# Multi Level Inheritance
# Definition: Multilevel inheritance is when a class inherits from another class, and a third class inherits from that child class.

"""
Grandparent
    ↓
  Parent
    ↓
  Child
  
"""  

class Campus(Reebok):
    def __init__(self, materials, zips, pockets, colours, size):
        super().__init__(materials, zips, pockets, colours)
        self.size = size

    def details(self):
        print("Size: ", self.size)
        return super().details()   

bag3 = Campus("Polyster", 1, 1, "Black", "XL")  
bag3.details()   
```

```python
# Multiple Inheritance 
# Definition: Multiple inheritance is when one child class inherits from two or more parent classes.

class ShoeFactory:
    def __init__(self, stuff, size, colour, brand):
        self.stuff = stuff
        self.size = size
        self.colour = colour
        self.brand = brand

    def details(self):
        print("Material: ", self.stuff)
        print("Size: ", self.size)
        print("Colour: ", self.colour)
        print("Brand: ", self.brand)    

class Customer(BagFactory, ShoeFactory):
    def __init__(self, name):
        BagFactory.__init__(self, "Leather", 2, 2)  
        ShoeFactory.__init__(self, "Nylon", 10, "Brown", "Bata")    
        self.name = name

    def details(self):
        print ("Customer name is: ", self.name)
        BagFactory.details(self)
        ShoeFactory.details(self)

customer1 = Customer("Lakshay")    
customer1.details()

"""

customer1.details()
       ↓
Customer.details()
       ↓
Print customer name
       ↓
BagFactory.details(self)
       ↓
Print bag details
       ↓
ShoeFactory.details(self)
       ↓
Print shoe details

"""
```
