```python
class Car:
    print("Hello How are you?")

# There are 2 types of things we store inside class:
# 1) Attributes - Variables defined inside class are attributes
# 2) Methods - Functions defined inside a class are methods.

class Animal:
    species = "Dog"   # Attribute

    def make_sound(self):   # Method
        print("Bark!")

# To access attributes and methods we have to first access the class
# and then attributes and methods

# Directly accessing attribute and method using the class

print(Animal.species)   # Accessing Attribute
Animal().make_sound()  # Call method
```
