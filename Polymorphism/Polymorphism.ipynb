```python
def hello():
    print("How are you?")

def hello():
    print("You are new to this.")    

hello()    # The second function replaces/reassigns the name hello,
           # so calling hello() runs the second function.
```

```python
class Animal:
    def speak(self):
        print("Animals will not speak.")

class Humans:
    def speak(self):
        print("We are humans.\nWe will not stay quiet.")     

obj = Animal()
obj1 = Humans()           

obj.speak()
obj1.speak()
```

```python
# Method Overriding (We need inheritance)

class Animal:
    def __init__(self, name):
        self.name = name

    def details(self):
        print(f"Your name is {self.name}")  

class Humans(Animal):
    def __init__(self, name):
        super().__init__(name)  

    def details(self):
        print(f"Your info is {self.name} and this is all we have.")                

obj = Humans("Amritanshu")     

obj.details()

# When we are doing inheritance and parent and child classes have same method name
# so the child class method will override your parent class method.
```
